```python
# generate_posters_fixed.py
from PIL import Image, ImageDraw, ImageFont, ImageFilter, ImageColor
import textwrap
import os

# ---------- CONFIG ----------
OUTPUT_DIR = "posters_output"
os.makedirs(OUTPUT_DIR, exist_ok=True)

# Put full paths to .ttf files if you want custom fonts; otherwise Pillow tries DejaVu.
FONT_PATH_BOLD = None  # e.g. "/usr/share/fonts/truetype/dejavu/DejaVuSans-Bold.ttf"
FONT_PATH_REG = None   # e.g. "/usr/share/fonts/truetype/dejavu/DejaVuSans.ttf"

# Poster size (px)
W, H = 1200, 1600

# Personal info (edit)
author_name = "Kaila Naresh"
role_line = "Aspiring Data Scientist • Python | ML | AI"
projects_short = "House Price Prediction • Churn Analysis • BERT Chatbot • Object Detection"

# Poster templates (title + paragraph)
posters = [
    {
        "title": "Data Science WordCloud Project",
        "paragraph": (
            "Unleashing the power of visualization! I created a dynamic Data Science WordCloud "
            "that highlights popular tools, techniques and concepts like Python, Pandas, "
            "Matplotlib and Seaborn. Visual storytelling makes complex ideas simple and memorable."
        ),
        "palette": ("#0f172a", "#0ea5a4")  # dark->teal
    },
    {
        "title": "Machine Learning & AI Portfolio",
        "paragraph": (
            "Building intelligent systems that create impact. From House Price Prediction and "
            "Customer Churn Analysis to Object Detection and BERT-based chatbots — each project "
            "focuses on accuracy, deployment-readiness, and real-world value."
        ),
        "palette": ("#071A52", "#6EE7B7")  # navy->mint
    },
    {
        "title": "Journey Into Data Science",
        "paragraph": (
            "Curiosity, consistency and creativity drive my data science journey. I continuously "
            "learn new tools and ship projects that decode data into actionable insights."
        ),
        "palette": ("#2a1a5e", "#f97316")  # purple -> orange
    }
]

# ---------- helper funcs ----------
def load_font(size, bold=False):
    """
    Tries to load configured TTF fonts. Falls back to DejaVu if available, otherwise to default bitmap font.
    """
    try:
        if bold and FONT_PATH_BOLD:
            return ImageFont.truetype(FONT_PATH_BOLD, size)
        if not bold and FONT_PATH_REG:
            return ImageFont.truetype(FONT_PATH_REG, size)
        # Try DejaVu shipped with many Pillow installs
        return ImageFont.truetype("DejaVuSans-Bold.ttf" if bold else "DejaVuSans.ttf", size)
    except Exception:
        # final fallback
        return ImageFont.load_default()

def make_vertical_gradient(w, h, start_color, end_color):
    """
    Create a vertical gradient image from start_color to end_color.
    """
    base = Image.new('RGB', (w, h), start_color)
    top = Image.new('RGB', (w, h), end_color)
    mask = Image.linear_gradient("L").resize((w, h))  # L mode mask (0->255)
    return Image.composite(top, base, mask)

def text_size(draw, text, font):
    """
    Return (width, height) using textbbox to support Pillow 10+.
    """
    bbox = draw.textbbox((0, 0), text, font=font)
    w = bbox[2] - bbox[0]
    h = bbox[3] - bbox[1]
    return w, h

# ---------- generate posters ----------
for i, p in enumerate(posters, start=1):
    # create gradient background
    bg = make_vertical_gradient(W, H, p["palette"][0], p["palette"][1])
    bg = bg.filter(ImageFilter.GaussianBlur(radius=3))
    draw = ImageDraw.Draw(bg)

    # fonts
    title_font = load_font(64, bold=True)
    subtitle_font = load_font(28)
    body_font = load_font(28)
    name_font = load_font(22, bold=True)

    # Title: centered near top
    title = p["title"]
    title_w, title_h = text_size(draw, title, title_font)
    draw.text(((W - title_w) / 2, 120), title, font=title_font, fill="white")

    # small divider line
    line_y = 120 + title_h + 20
    draw.line([(W * 0.15, line_y), (W * 0.85, line_y)], fill=(255, 255, 255, 180), width=3)

    # Paragraph: wrapped
    para = p["paragraph"]
    margin = 120
    max_width = W - 2 * margin
    # textwrap.fill is okay, but we want lines so use wrap()
    wrapped_lines = textwrap.wrap(para, width=40)
    y = line_y + 30
    for line in wrapped_lines:
        lw, lh = text_size(draw, line, body_font)
        draw.text(((W - lw) / 2, y), line, font=body_font, fill="white")
        y += lh + 8

    # Footer: author + role + projects
    footer_y = H - 220
    draw.text((margin, footer_y), author_name, font=name_font, fill="white")
    draw.text((margin, footer_y + 36), role_line, font=subtitle_font, fill="white")
    draw.text((margin, footer_y + 80), projects_short, font=subtitle_font, fill="white")

    # Save
    out_path = os.path.join(OUTPUT_DIR, f"poster_{i}.png")
    bg.save(out_path)
    print("Saved:", out_path)

print("All posters generated in:", OUTPUT_DIR)
```

```python
# interview_slide.py
from PIL import Image, ImageDraw, ImageFont, ImageFilter
import textwrap
import os

# ---------- CONFIG ----------
OUT = "interview_slide.png"
W, H = 1200, 675  # widescreen slide
BG1, BG2 = "#0b1220", "#0f172a"

# ---- Replace these with your project details ----
project_title = "House Price Prediction"
project_role = "End-to-end: Data + Model + Frontend (Streamlit)"
project_tech = "Python • Pandas • Scikit-learn • XGBoost • SHAP • Streamlit"
project_bullets = [
    "Cleaned dataset (30% missing) & engineered 12 new features (location, sqft per bed).",
    "Built XGBoost model with CV & early stopping; tuned hyperparams using randomized search.",
    "Added SHAP explanations for top features; exported model as pickle for Streamlit demo."
]
project_impact = "RMSE reduced from 55,000 → 28,000 (32% improvement). Interactive demo accepted by stakeholders."
# -------------------------------------------------

def load_font(sz, bold=False):
    try:
        return ImageFont.truetype("DejaVuSans-Bold.ttf" if bold else "DejaVuSans.ttf", sz)
    except:
        return ImageFont.load_default()

def gradient(w,h,c1,c2):
    base = Image.new("RGB",(w,h),c1)
    top = Image.new("RGB",(w,h),c2)
    mask = Image.linear_gradient("L").resize((w,h))
    return Image.composite(top, base, mask)

img = gradient(W,H,BG1,BG2).filter(ImageFilter.GaussianBlur(1))
draw = ImageDraw.Draw(img)

# Fonts
title_f = load_font(40, bold=True)
sub_f = load_font(20)
body_f = load_font(18)
impact_f = load_font(20, bold=True)

# Helpers
def tsize(draw, text, font):
    bbox = draw.textbbox((0,0), text, font=font)
    return bbox[2]-bbox[0], bbox[3]-bbox[1]

padding = 40
# Title block
draw.rectangle([padding, padding, W-padding, 140], fill=(255,255,255,20))
tw, th = tsize(draw, project_title, title_f)
draw.text((padding+10, padding+12), project_title, font=title_f, fill="white")
draw.text((padding+10, padding+12+th+6), project_role, font=sub_f, fill="white")

# Tech line (right side)
tech_w, _ = tsize(draw, project_tech, sub_f)
draw.text((W - padding - tech_w - 10, padding+18), project_tech, font=sub_f, fill="white")

# Bullets area
bx, by = padding+20, 180
max_w = W - 2*padding - 40
for b in project_bullets:
    wrapped = textwrap.wrap("• " + b, width=80)
    for line in wrapped:
        lw, lh = tsize(draw, line, body_f)
        draw.text((bx, by), line, font=body_f, fill="white")
        by += lh + 6
    by += 6

# Divider
draw.line([(padding, by+10),(W-padding, by+10)], fill=(255,255,255,40), width=1)

# Impact (emphasize)
imp_y = by + 30
draw.text((padding+10, imp_y), "Impact:", font=impact_f, fill="white")
iw, ih = tsize(draw, project_impact, body_f)
draw.text((padding+110, imp_y+2), project_impact, font=body_f, fill="white")

# Footer: your name / contact (change as needed)
footer = "Kaila Naresh — Python | ML | AI  •  github.com/nareshkaila  •  linkedin.com/in/kaila-naresh"
fw, fh = tsize(draw, footer, sub_f)
draw.text(((W-fw)/2, H-40), footer, font=sub_f, fill="white")

img.save(OUT)
print("Saved:", OUT)

```
